```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import current_timestamp, lit, from_utc_timestamp
from delta.tables import DeltaTable
from pyspark.sql.utils import AnalysisException
import os

path = f"Tables/Run_Timestamp_Log"
ts = current_timestamp()

df = (
    spark.createDataFrame([(1,)], ["id"])
    .withColumn("last_run_timestamp_utc", ts)
    .withColumn("last_run_timestamp_est", from_utc_timestamp(ts, "America/New_York"))
    .withColumn("last_run_timestamp_ist",from_utc_timestamp(ts,"Asia/Kolkata"))
)
```
